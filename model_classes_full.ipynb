```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
import seaborn as sns
from matplotlib.image import imread
from PIL import Image
import tensorflow as tf
np.random.seed(1337)
import gc

from tensorflow.keras.utils import to_categorical
from tensorflow.keras import layers
from keras.callbacks import ReduceLROnPlateau
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Activation, Dropout, AveragePooling2D,Flatten, Dense, Conv2D,MaxPool2D, MaxPooling2D, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import MultiLabelBinarizer

img_size = 128
print(os.listdir())
dataset = os.listdir("music_dataset_spectro_full/train")
labels = dataset
print(labels)
```

```python
def get_dataset_array_train(data_dir):
    data = []
    for label in labels:
        print(label)
        path = os.path.join(data_dir, label)
        class_num = labels.index(label)
        for img in os.listdir(path):
            try:
                img_arr = cv2.imread(os.path.join(path, img),0)
                resized_arr = cv2.resize(img_arr, (img_size, img_size))
                data.append([resized_arr, [class_num]])
                gc.collect()
            except Exception as e:
                print(e)

    path = os.path.join("music_separation_2/train/"+"full_mix")
    for img in os.listdir("music_separation_2/train/"+"full_mix"):
        class_num = []
        for stem in os.listdir("music_separation_2/train/"+"stems"):
            if img.split("_")[0] == stem.split("_")[0]:
                temp=stem.split("_",1)
                label = temp[1].split(".")[0] 
                class_num.append(labels.index(label))

        try:
            img_arr = cv2.imread(os.path.join(path,img),0)
            resized_arr = cv2.resize(img_arr, (img_size, img_size))
            data.append([resized_arr,class_num])
            gc.collect()
        except Exception as e:
                print(e)
    return np.array(data,dtype="object")
```

```python
def get_dataset_array_test(data_dir):
    data = []
    for label in labels:
        print(label)
        path = os.path.join(data_dir, label)
        class_num = labels.index(label)
        for img in os.listdir(path):
            try:
                img_arr = cv2.imread(os.path.join(path, img),0)
                resized_arr = cv2.resize(img_arr, (img_size, img_size))
                data.append([resized_arr, [class_num]])
                gc.collect()
            except Exception as e:
                print(e)

    path = os.path.join("music_separation_2/test/"+"full_mix")
    for img in os.listdir("music_separation_2/test/"+"full_mix"):
        class_num = []
        for stem in os.listdir("music_separation_2/test/"+"stems"):
            if img.split("_")[0] == stem.split("_")[0]:
                temp=stem.split("_",1)
                label = temp[1].split(".")[0] 
                class_num.append(labels.index(label))

        try:
            img_arr = cv2.imread(os.path.join(path,img),0)
            resized_arr = cv2.resize(img_arr, (img_size, img_size))
            data.append([resized_arr,class_num])
            gc.collect()
        except Exception as e:
                print(e)
    return np.array(data,dtype="object")
```

```python
def get_dataset_array_valid(data_dir):
    data = []
    for label in labels:
        print(label)
        path = os.path.join(data_dir, label)
        class_num = labels.index(label)
        for img in os.listdir(path):
            try:
                img_arr = cv2.imread(os.path.join(path, img),0)
                resized_arr = cv2.resize(img_arr, (img_size, img_size))
                data.append([resized_arr, [class_num]])
                gc.collect()
            except Exception as e:
                print(e)

    path = os.path.join("music_separation_2/valid/"+"full_mix")
    for img in os.listdir("music_separation_2/valid/"+"full_mix"):
        class_num = []
        for stem in os.listdir("music_separation_2/valid/"+"stems"):
            if img.split("_")[0] == stem.split("_")[0]:
                temp=stem.split("_",1)
                label = temp[1].split(".")[0] 
                class_num.append(labels.index(label))

        try:
            img_arr = cv2.imread(os.path.join(path,img),0)
            resized_arr = cv2.resize(img_arr, (img_size, img_size))
            data.append([resized_arr,class_num])
            gc.collect()
        except Exception as e:
                print(e)
    return np.array(data,dtype="object")
```

```python
train = get_dataset_array_train("music_dataset_spectro_full/train/")
test = get_dataset_array_test("music_dataset_spectro_full/test/")
valid = get_dataset_array_valid("music_dataset_spectro_full/valid/")
```

```python
print(train.shape)
print(test.shape)
print(valid.shape)
```

```python
x_train = []
y_train = []

x_val = []
y_val = []

x_test = []
y_test = []

for feature, label in train:
    x_train.append(feature)
    y_train.append(label)

for feature, label in test:
    x_test.append(feature)
    y_test.append(label)

for feature, label in valid:
    x_val.append(feature)
    y_val.append(label)
     
del train
del test
del valid
```

```python
gc.collect()
x_train = np.array(x_train)/255
gc.collect()
x_test = np.array(x_test)/255
gc.collect()
x_val = np.array(x_val)/255
gc.collect()
```

```python
x_train = x_train.reshape(-1, img_size, img_size, 1)

mlb = MultiLabelBinarizer()

print(y_train)
y_train = mlb.fit_transform(y_train)

x_val = x_val.reshape(-1, img_size, img_size, 1)

y_val = mlb.fit_transform(y_val)


x_test = x_test.reshape(-1, img_size, img_size, 1)

y_test = mlb.fit_transform(y_test)
```

```python
# Model setup
model = Sequential()

model.add(Conv2D(32, (3,3), activation = 'relu', input_shape = (img_size, img_size, 1)))
model.add(MaxPool2D((2,2)))
model.add(BatchNormalization())

model.add(Conv2D(64, (3,3), activation = 'relu', input_shape = (img_size, img_size, 1)))
model.add(MaxPool2D((2,2)))
model.add(BatchNormalization())

model.add(Conv2D(128, (3,3), activation = 'relu', input_shape = (img_size, img_size, 1)))
model.add(MaxPool2D((2,2)))
model.add(BatchNormalization())

model.add(Flatten())
model.add(Dense(units = 128, activation = 'relu'))
model.add(Dropout(0.5))
model.add(Dense(units = 24, activation = 'sigmoid'))

model.compile(
              optimizer = 'adam', loss = 'binary_crossentropy',
              metrics = ['accuracy']
              )
     
```

```python
model.summary()
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
learning_rate_reduction = ReduceLROnPlateau(monitor = 'val_accuracy', patience = 2, verbose = 1, factor = 0.3, min_lr = 0.000001)
```

```python
batch_size = 32
n_epochs = 50
model.fit(x_train, y_train, batch_size = batch_size,
                    epochs = n_epochs, validation_data = (x_val, y_val),
                    callbacks = [learning_rate_reduction])
```

```python
gc.collect()
print("Accuracy of the model is - " , model.evaluate(x_test,y_test)[1]*100 , "%")
model.save('my_model_full.keras')
```
